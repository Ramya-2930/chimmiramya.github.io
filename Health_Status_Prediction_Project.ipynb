```
import numpy as np
# Recreate a LabelEncoder specifically for the 'General_Health' target variable.
# This is necessary because the global 'le' object was re-fit on other binary columns
# and no longer holds the classes for 'General_Health'.
# The original unique values for 'General_Health' were ['Poor', 'Very Good', 'Good', 'Fair', 'Excellent'].
# LabelEncoder maps these alphabetically, so the sorted classes are:
# ['Excellent', 'Fair', 'Good', 'Poor', 'Very Good']
le_general_health_reconstructed = LabelEncoder()
le_general_health_reconstructed.fit(np.array(['Excellent', 'Fair', 'Good', 'Poor', 'Very Good'], dtype=object))

health_status = le_general_health_reconstructed.inverse_transform(prediction)
print("===================================")
print("GENERAL HEALTH PREDICTION SYSTEM")
print("===================================")
print("Prediction :", health_status[0])
print("===================================")
```
